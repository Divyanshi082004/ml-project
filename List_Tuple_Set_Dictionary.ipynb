List

```python
#list should be included in the square brackets
my_list=[1,2,3,4,5]
print(my_list)
type(my_list)
```

```python
# list can have multiple data types
my_list=[2,3,'divi',1.5]
print(my_list)
type(my_list)
type(1.5)
```

lists are mutable-->Changeable

```python
#add elements to a list
my_list = [1,2,3.5,'divi',True]
my_list.append(8)
print(my_list)
```

```python
#print elements of a list using their index
print(my_list[2])
print(my_list[5])
```

```python
#list allow duplicates values
list_1 =[1,2,3,4,5,12,2,3,39,4]
print(list_1)
#check the lenth of list
print(len(list_1))
```

```python
#initiating an empty list
list_2 = []
print(list_2)
```

```python
list_2.append(1)
list_2.append(2)
list_2.append(3)
list_2.append(4)
print(list_2)
```

```python
#delete the item from list
del list_2[2]
print(list_2)
```

```python
#join two lists
list_3 = list_1 + list_2
print(list_3)
```

Tuple

```python
tuple_1 = (2,3,4,5,6)
print(tuple_1)
type(tuple_1)
```

```python
#tuple allows multiple data types
tuple_2 = (2,3,4.5,'divi',True)
print(tuple_2)
```

```python
#converting a list into a tuple
my_list = [3,4,5,6]
print(my_list)
my_tuple = tuple(my_list)
print(my_tuple)
```

```python
print(my_tuple[0])
print(my_tuple[3])
```

Tuple are immutable-->Unchangeable

```python
#it give error because tuple are unchangeable
my_tuple.append(2)
print(my_tuple)
```

```python
print(len(my_tuple))
```

Set


```python
# set --->Curly brackets
my_set = {1,2,3,4,5}
print(my_set)
type(my_set)
```

```python
#set does not supports indexing
print(my_set[0])
```

```python
#convert list into set
list_4 =[4,6,90,10,30,20]
x = set(list_4)
print(x)
```

```python
#set does not allow duplicate values
set_3 = {1,2,3,4,5,6,2,7,8}
print(set_3)
```

Dictionary

```python
#it have key value pair
my_dic = {'name':'divi','age':20,'country':'india'}
print(my_dic)
type(my_dic)
```

```python
print(my_dic['name'])
print(my_dic['age'])
print(my_dic['country'])
```
