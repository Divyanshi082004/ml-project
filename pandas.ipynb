pivot() ka kaam kya hota hai?

👉 Rows ko columns me badal deta hai
👉 Ek column ki unique values ko naya column bana deta hai
👉 Table ko reshape karta hai
✅ index='name'
→ name column row index ban jayega

✅ columns='age'
→ age column ki unique values (jaise 20, 21, 22) nayi columns ban jayengi

✅ values='Surname'
→ cells ke andar Surname ki values fill hongi


```python
df.pivot(index='name', columns='age', values='Surname')
```
