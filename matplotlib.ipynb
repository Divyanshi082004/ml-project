Bar plot

```python
fir = plt.figure()
ax = fir.add_axes([0,0,1,1])
language = ['English','hindi','Marathi','Gujarati','Spanish']
people = [100,150,50,40,70]
ax.bar(language,people)
plt.xlabel('language')
plt.ylabel('people')
plt.show()
```

Pie chart

```python
fig1 = plt.figure()
ax = fig1.add_axes([0,0,1,1])
language = ['English','hindi','Marathi','Gujarati','Spanish']
people = [100,150,50,40,70]
ax.pie(people, labels = language, autopct='%1.1f%%')
plt.show()
```

Scatter plot

```python
x = np.linspace(0,10,30)
y = np.sin(x)
z = np.cos(x)
fig2 = plt.figure()
ax = fig2.add_axes([0,0,1,1])
ax.scatter(x,y,color='red')
ax.scatter(x,z,color='blue')
plt.show()
```

3D scatter graph

```python
fig3 = plt.figure()
ax = plt.axes(projection='3d')
z = 20 * np.random.random(100)
x = np.sin(z)
y = np.cos(z)
ax.scatter3D(x,y,z,c=z,cmap='Greens')
plt.show()
```
